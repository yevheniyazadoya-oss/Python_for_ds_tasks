Будемо працювати з даними з файлу `cookie_cats.csv`. Колонки в даних наступні:

Коли гравець встановлював гру, його випадковим чином призначали до групи gate_30 або gate_40.

1. Для початку, уявімо, що ми тільки плануємо проведення зазначеного А/B-тесту і хочемо зрозуміти, дані про скількох користувачів нам треба зібрати, аби досягнути відчутного ефекту. Відчутним ефектом ми вважатимемо збільшення утримання на 1% після внесення зміни. Обчисліть, скільки користувачів сумарно нам треба аби досягнути такого ефекту, якщо продакт менеджер нам повідомив, що базове утримання є 19%.

````python
effect_size = sms.proportion_effectsize(0.20, 0.19)
````

````python
required_n = sms.NormalIndPower().solve_power(
    effect_size,
    power=0.8,
    alpha=0.05,
    ratio=1
    )                                                  # Розрахунок необхідного розміру вибірки

required_n = ceil(required_n)                          # Округлення до наступного цілого числа

print(required_n)
````

````python
control_sample = df[df['version'] == 'gate_30'].sample(n=required_n, random_state=22)
treatment_sample = df[df['version'] == 'gate_40'].sample(n=required_n, random_state=22)

ab_test = pd.concat([control_sample, treatment_sample], axis=0)
ab_test.reset_index(drop=True, inplace=True)
````

````python
ab_test.head()
````

````python
ab_test['version'].value_counts()
````

Тестування гіпотези:

````python
from statsmodels.stats.proportion import proportions_ztest, proportion_confint
````

````python
control_results = ab_test[ab_test['version'] == 'gate_30']['retention_7']
treatment_results = ab_test[ab_test['version'] == 'gate_40']['retention_7']
````

````python
n_con = control_results.count()
n_treat = treatment_results.count()
successes = [control_results.sum(), treatment_results.sum()]
nobs = [n_con, n_treat]
````

````python
successes
````

````python
z_stat, pval = proportions_ztest(successes, nobs=nobs)
(lower_con, lower_treat), (upper_con, upper_treat) = proportion_confint(successes, nobs=nobs, alpha=0.05)

print(f'z statistic: {z_stat:.2f}')
print(f'p-value: {pval:.3f}')
print(f'Довірчий інтервал 95% для групи control: [{lower_con:.3f}, {upper_con:.3f}]')
print(f'Довірчий інтервал 95% для групи treatment: [{lower_treat:.3f}, {upper_treat:.3f}]')
````

**Відповіді на питання.**

1. Чи є різниця статистично значущою?

Так. Z-тест для двох пропорцій дав z = 3.20 і p-value = 0.001, що менше за 0.05. Тому нульову гіпотезу про однакове утримання на 7-й день в обох версіях відхиляємо.

2. Чи перетинаються довірчі інтервали? Про що це каже?

Ні, не перетинаються: для control (gate_30) інтервал [0.184, 0.194], для treatment (gate_40) - [0.173, 0.183]. Верхня межа інтервалу treatment (0.183) нижча за нижню межу інтервалу control (0.184).

Це підтверджує результат z-тесту: різниця між версіями не пояснюється випадковими коливаннями вибірки, а є реальним ефектом.

4. Виконайте тест Хі-квадрат на рівні значущості 5% аби визначити, чи є залежність між версією гри та утриманням гравця на 7ий день після реєстрації.

    - Напишіть, як для цього тесту будуть сформульовані гіпотези.
    - Проведіть обчислення, виведіть p-значення і напишіть висновок за результатами тесту.


**Гіпотези:**

* H₀: версія гри та утримання на 7-й день незалежні (розподіл retention_7

однаковий в обох версіях).

* H₁: між версією гри та утриманням на 7-й день є залежність.

````python
pd.crosstab(ab_test['version'], ab_test['retention_7'])
````

````python
crosstab = pd.crosstab(ab_test['version'], ab_test['retention_7'])
````

````python
chi2, p, dof, expected = stats.chi2_contingency(crosstab)

print(f"χ² = {chi2:.3f}")
print(f"p-value = {p:.5f}")
print(f"Ступені свободи = {dof}")
print("Очікувані частоти:\n", expected)
````

**Висновок:**

p-value = 0.00143 < 0.05, тому на рівні значущості 5% відхиляємо нульову гіпотезу. Між версією гри та утриманням гравців на 7-й день є статистично значуща залежність: частка гравців, що повернулися через 7 днів, відрізняється залежно від того, на якому рівні стоять ворота (30 чи 40).

Отже, обидва тести дають однаковий результат: перенесення воріт на 40-й рівень погіршує утримання на 7-й день, тому доцільніше залишити ворота на 30-му рівні.
