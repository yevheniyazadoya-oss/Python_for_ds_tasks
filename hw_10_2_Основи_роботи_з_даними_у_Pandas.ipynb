**Перевірка третьої гіпотези**

Судячи з розрахунку середнього значення інвойсу між чоловіками та жінками спостерігається в двох містах тенденція, що жінки приносять більше валового доходу ніж чоловіки, але в місті Mandalay середнє значення 1 інвойсу жінками трохи нижче ніж в місті Yangon.
Гіпотеза підтверджена.

```python
filter_city = df.City == 'Yangon'
filter_female = df.Gender == 'Female'
filter_male = df.Gender == 'Male'

female_income = df[filter_city & filter_female]['gross income'].mean()
male_income = df[filter_city & filter_male]['gross income'].mean()

print(f'Gross income from female:{female_income}, Gross income from male: {male_income}')
```

```python
filter_city = df.City == 'Mandalay'

print(f'Gross income from female:{female_income}, Gross income from male: {male_income}')
```
