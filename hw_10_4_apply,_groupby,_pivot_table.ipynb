**1. Чи залежить середній чек від типу покупця по різним категоріям товарів?**

Згідно отриманих результатів спостерігається ситуація, що середній чек по всім категоріям товарів окрім "Sports and travel" вищий по типу покупця "Member" ніж по "Normal". Велика різниця спостерігається між типами покупців по середньому чеку в категорії "Health and beauty". Дослідивши статистичні показники по кожному типу покупців можна зробити висновок, що у категорії "Health and beauty" покупці категорії Member демонструють систематично вищий чек порівняно зі звичайними покупцями Normal.

Середнє значення по двом типам покупців суттєво вище за медіану — це означає, що розподіл витягнутий вправо (є частина покупок з високим чеком, які "тягнуть" середнє вгору). Розрив зберігається і на 25-му, і на 75-му перцентилі між типами покупців з чого можна зробити висновок, що Member-покупці стабільно витрачають більше за Normal-покупців в цій категорії на всіх рівнях вартості покупки, а не тільки за рахунок outliers.

**Рекомендація для бізнесу:** оскільки Member вже органічно витрачають більше в цій категорії, варто розглянути персоналізовані пропозиції в категорії Health and beauty для учасників програми лояльності — там найбільший потенціал для крос-продажу та збільшення частоти покупок.

```python
df.pivot_table(index='Product line', columns='Customer type', values='Total', aggfunc='mean').round(2)
```

```python
df_selected_cols = df[['Customer type', 'Total', 'Gender', 'Payment', 'Product line']]
```

```python
# статитистичні дані по категорії "Health and beauty" по типу покупця "Member"
df_selected_cols[
    (df_selected_cols['Product line'] == 'Health and beauty') & (df_selected_cols['Customer type'] == 'Member')].describe(percentiles=[0.01, 0.25, 0.50, 0.75, 0.99])
```

```python
# статитистичні дані по категорії "Health and beauty" по типу покупця "Normal"
df_selected_cols[
    (df_selected_cols['Product line'] == 'Health and beauty') & (df_selected_cols['Customer type'] == 'Normal')].describe(percentiles=[0.01, 0.25, 0.50, 0.75, 0.99])
```

**2. Чи є залежність суми середнього чеку від гендеру та типу оплати?**

Згідно отриманих результатів можна спостерігати, що є залежність від гендеру, де видно, що жінки обирають дорожчі товари ніж чоловіки при незначній перевищеній середній кількості товару в чеку по всім способам оплат.

Credit card — спосіб оплати з найменшою гендерною різницею (+2.8%), де поведінка чоловіків і жінок схожа.

У способі оплати Ewallet жінки демонструють суттєво вищий чек порівняно з чоловіками: медіана перевищує чоловічу на 46.5% (299.89 проти 204.70). Це вказує на стійку та суттєву відмінність купівельної поведінки за гендером саме в цьому каналі оплати.

**Рекомендації для бізнесу:** перевірити гіпотезу, що жінки частіше використовують Ewallet саме для великих/планових покупок, а чоловіки — для дрібних/спонтанних; розглянути впровадження промо для способу оплати Ewallet для жіночої аудиторії.

```python
df.pivot_table(index='Payment', columns='Gender', values=['Total', 'Quantity'], aggfunc='mean').round(2)
```

```python
# статитистичні дані по способу оплати "Ewallet" по гендеру "Female"
df_selected_cols[
    (df_selected_cols['Payment'] == 'Ewallet') & (df_selected_cols['Gender'] == 'Female')].describe(percentiles=[0.01, 0.25, 0.50, 0.75, 0.99])
```

```python
# статитистичні дані по способу оплати "Ewallet" по гендеру "Male"
df_selected_cols[
    (df_selected_cols['Payment'] == 'Ewallet') & (df_selected_cols['Gender'] == 'Male')].describe(percentiles=[0.01, 0.25, 0.50, 0.75, 0.99])
```
