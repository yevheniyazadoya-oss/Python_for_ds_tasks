**Розв'язок:**
1. Параметри задачі у змінних Python:

```python
sample_mean = np.mean(data)           # x¯, середнє значення в вибірці
population_mean = 20                  # μ0 - середнє значення в популяції
population_std = np.std(data, ddof=1) # - σ, вибіркове стандартне відхилення
sample_size = len(data)               # n - розмір вибірки
alpha = 0.05                          # α - рівень значущості
```

2. Гіпотези:

H₀: μ = 20 (середній чек складає 20$)

H₁: μ > 20 (справжній середній чек за липень 2025 перевищує попередні 20$)

Тип тесту: правосторонній (one-tailed)

3. Обчислення стандартної помилки SE:

```python
# Формула: SE = σ / √n
standard_error = population_std / np.sqrt(sample_size)

print(f"SE = {population_std} / √{sample_size}")
print(f"SE = {population_std} / {np.sqrt(sample_size):.3f}")
print(f"SE = {standard_error:.3f}")
print()
print("Інтерпретація: В середньому, вибіркове середнє відхиляється")
print(f"від справжнього середнього на ±{standard_error:.3f}$")
```

4. Розрахунок z-статистики:

```python
# Формула: z = (x̄ - μ₀) / SE
z_statistic = (sample_mean - population_mean) / standard_error

print(f"z = ({sample_mean:.3f} - {population_mean}) / {standard_error:.3f}")
print(f"z = {sample_mean - population_mean:.3f} / {standard_error:.3f}")
print(f"z = {z_statistic:.3f}")
print()
print("Інтерпретація:")

direction = "вище" if z_statistic > 0 else "нижче"
print(f"Наше вибіркове середнє на {abs(z_statistic):.3f} стандартних "
      f"помилок {direction} за гіпотетичне середнє популяції.")
print()

if abs(z_statistic) > 2:
    print("Це досить велике відхилення (більше 2 стандартних помилок).")
elif abs(z_statistic) > 1:
    print("Це помірне відхилення (між 1 та 2 стандартними помилками).")
else:
    print("Це невелике відхилення (менше 1 стандартної помилки).")
```

5. Знаходження критичного z-значення за допомогою бібліотеки stats:

```python
# Для правостороннього тесту з α = 0.05
z_critical = stats.norm.ppf(1 - alpha)

print(f"Для рівня значущості α = {alpha} (правосторонній тест)")
print(f"Критичне z-значення = {z_critical:.3f}")
```

6. Обчислення p-value з допомогою бібліотеки stats:

```python
# Для правостороннього тесту: p-value = P(Z > z_statistic)
p_value = 1 - stats.norm.cdf(z_statistic)

print(f"p-value = {p_value:.6f}")
print()
print("Що це означає?")
print(f"Якщо H₀ правильна (середній чек не більше 20$), то ймовірність")
print(f"випадково отримати середній чек ≥ {sample_mean} становить {p_value:.6f}")
print(f"або приблизно {p_value*100:.4f}%")
print()
if p_value < 0.001:
    print("Це НАДЗВИЧАЙНО малоймовірно! (менше 0.1%)")
elif p_value < 0.01:
    print("Це дуже малоймовірно! (менше 1%)")
elif p_value < 0.05:
    print("Це малоймовірно! (менше 5%)")
else:
    print("Це цілком можливо випадково")
```

7. Прийміть рішення, чи відхиляєте ви гіпотезу  H0 . Для прийняття рішення зробіть і порівняння z-статистики з критичним значенням, і проаналізуйте p-value.

```python
# Метод 1: Порівняння z-статистики з критичним значенням

print(f"z-статистика = {z_statistic:.3f}")
print(f"Критичне значення = {z_critical:.3f}")
print(f"Порівняння: {z_statistic:.3f} {'>' if z_statistic > z_critical else '≤'} {z_critical:.3f}")

if z_statistic > z_critical:
    print("Висновок: z-статистика ПЕРЕВИЩУЄ критичне значення")
    decision1 = "ВІДХИЛЯЄМО H₀"
else:
    print("Висновок: z-статистика НЕ перевищує критичне значення")
    decision1 = "НЕ ВІДХИЛЯЄМО H₀"

print(f"Рішення: {decision1}")
```

```python
# Метод 2: Порівняння p-value з рівнем значущості

print(f"p-value = {p_value:.6f}")
print(f"Рівень значущості α = {alpha}")
print(f"Порівняння: {p_value:.6f} {'<' if p_value < alpha else '≥'} {alpha}")

if p_value < alpha:
    print("Висновок: p-value МЕНШЕ за рівень значущості")
    decision2 = "ВІДХИЛЯЄМО H₀"
else:
    print("Висновок: p-value НЕ менше за рівень значущості")
    decision2 = "НЕ ВІДХИЛЯЄМО H₀"

print(f"Рішення: {decision2}")
```

```python
# Фінальний висновок:
assert decision1 == decision2, "Методи дали різні рішення, перевірте обчислення!"

if decision2 == "ВІДХИЛЯЄМО H₀":
    print("✅ Середній чек статистично значуще ПІДВИЩИВСЯ!")
    print(f"   На рівні значущості {alpha:.0%} середній чек за липень 2025")
    print(f"   ({sample_mean:.2f}$) статистично значуще перевищує")
    print(f"   попереднє значення ({population_mean}$).")
else:
    print("❌ Підвищення середнього чека НЕ ПІДТВЕРДЖУЄТЬСЯ")
    print(f"   На рівні значущості {alpha:.0%} немає достатніх статистичних")
    print("   доказів, що середній чек за липень 2025 перевищує")
    print(f"   попередні {population_mean}$.")
    print(f"   Вибіркове середнє ({sample_mean:.2f}$) навіть трохи нижче за {population_mean}$,")
    print(f"   а p-value = {p_value:.3f} значно більше за α = {alpha}.")
```

```python
# Виконуємо правосторонній z-тест
z_stat_sm, p_val_sm = stests.ztest(data,           # наші дані
                                    value=population_mean,  # μ₀ = 20
                                    alternative='larger')   # H₁: μ > 20

print(f"Результати z-тесту:")
print(f"  Z-статистика: {z_stat_sm:.3f}")
print(f"  P-value: {p_val_sm:.6f}")
print()

# Інтерпретація результатів
print("ІНТЕРПРЕТАЦІЯ:")
print("-" * 40)
if p_val_sm < alpha:
    print(f"✅ ВІДХИЛЯЄМО H₀")
    print(f"   p-value ({p_val_sm:.6f}) < α ({alpha})")
    print("   Висновок: Середній чек за липень 2025 статистично значуще перевищує 20$")
else:
    print(f"❌ НЕ ВІДХИЛЯЄМО H₀")
    print(f"   p-value ({p_val_sm:.6f}) ≥ α ({alpha})")
    print("   Висновок: Немає достатніх доказів про вищий середній чек")

# Порівняння з ручними розрахунками
print("\n" + "=" * 60)
print("ПОРІВНЯННЯ З РУЧНИМИ РОЗРАХУНКАМИ:")
print("=" * 60)
print(f"Z-статистика:")
print(f"  Ручний розрахунок: {z_statistic:.3f}")
print(f"  Statsmodels:       {z_stat_sm:.3f}")
print(f"  Різниця:           {abs(z_statistic - z_stat_sm):.3f}")
print()
print("P-value:")
print(f"  Ручний розрахунок: {p_value:.6f}")
print(f"  Statsmodels:       {p_val_sm:.6f}")
print(f"  Різниця:           {abs(p_value - p_val_sm):.6f}")
```
